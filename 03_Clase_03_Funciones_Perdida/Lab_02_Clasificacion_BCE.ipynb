## 1. Sigmoide
Transforma logits reales en probabilidades.

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
z = np.linspace(-6,6,200)
plt.plot(z,sigmoid(z)); plt.xlabel("Logit");plt.ylabel("Probabilidad");plt.grid(alpha=.2);plt.show()
```

## 3. Penalización por confianza
Compare la pérdida para una etiqueta y=1 y distintas probabilidades.

```python
probs = np.linspace(.01,.99,100)
loss = -np.log(probs)
plt.plot(probs,loss);plt.xlabel("P(y=1)");plt.ylabel("−log P(y=1)");plt.grid(alpha=.2);plt.show()
```
